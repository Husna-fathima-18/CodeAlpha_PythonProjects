```python
import os
import shutil

# Define folders
source_folder = "source_folder"
destination_folder = "jpg_files"

print("=" * 50)
print("       AUTOMATED JPG FILE ORGANIZER")
print("=" * 50)

# Check source folder
if not os.path.exists(source_folder):
    print("Source folder does not exist.")
    print("Please create a folder named 'source_folder'.")
else:

    # Create destination folder if it doesn't exist
    if not os.path.exists(destination_folder):
        os.makedirs(destination_folder)

    moved_files = 0

    # Check files in source folder
    for filename in os.listdir(source_folder):

        source_path = os.path.join(source_folder, filename)

        # Check if the item is a JPG/JPEG file
        if os.path.isfile(source_path) and filename.lower().endswith((".jpg", ".jpeg")):

            destination_path = os.path.join(
                destination_folder,
                filename
            )

            # Move file
            shutil.move(source_path, destination_path)

            print("Moved:", filename)

            moved_files += 1

    print("\n" + "=" * 50)

    if moved_files == 0:
        print("No JPG/JPEG files were found.")
    else:
        print(f"{moved_files} JPG/JPEG file(s) moved successfully.")

    print("=" * 50)
```
